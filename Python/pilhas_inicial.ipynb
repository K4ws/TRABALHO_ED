# Implementação inicial da IA e os testes iniciais.

```python
from array import array

class PilhaCheiaErro(Exception):
    """Erro gerado quando a pilha está cheia."""
    pass

class PilhaVaziaErro(Exception):
    """Erro gerado quando a pilha está vazia."""
    pass

class TipoErro(Exception):
    """Erro gerado quando o tipo do dado é incompatível."""
    pass

# --- IMPLEMENTAÇÃO INICIAL DA IA (COM FALHAS) ---
class Pilha:
    def __init__(self, capacidade):
        self.capacidade = capacidade
        self.dados = []  # FALHA: Usou lista nativa dinâmica em vez do módulo array

    def empilha(self, dado):
        if len(self.dados) < self.capacidade:
            self.dados.append(dado)  # FALHA: Crescimento dinâmico
        else:
            raise PilhaCheiaErro("A pilha está cheia.")

    def desempilha(self):
        if len(self.dados) == 0:
            raise PilhaVaziaErro("A pilha está vazia.")
        return self.dados.pop()  # FALHA: Redimensionamento em tempo de execução

    def pilha_esta_vazia(self):
        return len(self.dados) == 0

    def pilha_esta_cheia(self):
        return len(self.dados) == self.capacidade

    def troca(self):
        if len(self.dados) >= 2:
            self.dados[-1], self.dados[-2] = self.dados[-2], self.dados[-1]

    def tamanho(self):
        return len(self.dados)

# --- TESTE INICIAL ---
if __name__ == "__main__":
    print("--- Executando Teste Inicial ---")
    pilha = Pilha(3)
    pilha.empilha(10)
    pilha.empilha(20)
    pilha.empilha(30)

    print("Dados:", pilha.dados)
    print("Tamanho:", pilha.tamanho())
    print("Está cheia?", pilha.pilha_esta_cheia())
    print("Elemento removido:", pilha.desempilha())
    print("Tamanho após remoção:", pilha.tamanho())
```
